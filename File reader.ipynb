This piece of code is basically a practice for executing the tasks on a txt file. These are really easy for a txt file.

```python
from pypdf import PdfReader

txt=''

for f in p.iterdir():
    if f.suffix==".pdf":
        print(f)
        reader=PdfReader(f)
        count=0
        for page in reader.pages:
            count+=len(page.extract_text().split())

        print(f.name)
        print("Word Count : ",count)
```

Okay, this was the code for finding name and word count for pdfs, now let's merge the codes for both pdf and text.

There are also a few extra lines of code that i can remove, for eg: i did not need txt variable, i just used it to see the output while trying out the code, but now that i understand, i don't need it.

```python
from pathlib import Path
from pypdf import PdfReader

p=Path.cwd()
for files in p.iterdir():
    if files.suffix==".txt":
        print(files.name)
        print("Word Count : ",len(open(files).read().split()))
        print("File Type : .txt")
    elif files.suffix==".pdf":
        print(files.name)
        count=0
        for pages in PdfReader(files).pages:
            text = pages.extract_text() or ""
            count += len(text.split())

        print("Word Count : ",count)
        print("File Type : .pdf")
```
