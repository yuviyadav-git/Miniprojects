An Oop based project with a `Book`, `Member`, and `Library` class with borrowing/returning logic, inventory tracking, and basic error handling.

```python
class Book:
    def __init__(self,name,author,book_id,copy_id):
        self.name=name
        self.author=author
        self.book_id=book_id
        self.copy_id=copy_id
        self.borrowed=False
        self.borrower=None

class Member:
    def __init__(self,name,mem_id):
        self.name=name
        self.mem_id=mem_id
        self.borrowed=[]

class Library:
    def __init__(self):
        self.books=[]
        self.members=[]

    def add_book(self,newBook):
        for book in self.books:
            if book.book_id==newBook.book_id:
                if book.name.lower()!=newBook.name.lower():
                    print("Another title with the same book id exists\nGive a different id to this book.")
                    return
                elif book.copy_id==newBook.copy_id:
                    print("No two books with same Book id and Copy id can exist.")
                    return
                elif book.author!=newBook.author:
                    print("Same title cannot have different authors.")
                    return
            elif book.name.lower()==newBook.name.lower():
                print("Same title cannot have two different book IDs.")
                return
        self.books.append(newBook)
            

    def del_book(self,oldBook):
        for book in self.books:
            if book.book_id==oldBook.book_id and book.copy_id==oldBook.copy_id:
                print(f"Removing this book.\nName : {book.name}\nAuthor : {book.author}\nBook ID : {book.book_id}\nCopy ID : {book.copy_id}\n\nREMOVAL COMPLETE.")
                return

        print("Book not found.")

    def add_member(self,newMember):
        for member in self.members:
            if member.mem_id==newMember.mem_id:
                print("A member with this ID already exists.")
                return

        self.members.append(newMember)
        print("New member added")

    def del_member(self, oldMember):
        for member in self.members:
            if member.mem_id==oldMember.mem_id and member.name.lower()==oldMember.name.lower():
                print(f"Deleting {member.name}, Member ID : {member.mem_id}")
                print("Deletion Complete")
                return

        print("No such member found.")

    def borrow(self,bookName,borrower):
        temp=None
        for member in self.members:
            if member.name.lower()==borrower.name.lower() and member.mem_id==borrower.mem_id:
                print("Member verified.\nSearching for the book")
                temp=member
                break
        if temp is None:
            print(f"{borrower.name} is not a member. Only members can borrow a book.")
            return
        
        found=False
        for book in self.books: 
            if book.name.lower()==bookName.lower():
                found=True
                if  book.borrowed==False:
                    print(f"{bookName} found.")
                    temp.borrowed.append(book)
                    book.borrowed=True
                    book.borrower=temp
                    return

        if found:
            print(f"{bookName} found, but not available.")
        else:
            print("Book not found in Library.")

    def returnBook(self,bookName,borrower):
        temp=None
        for member in self.members:
            if member.name.lower()==borrower.name.lower() and member.id==borrower.id:
                print("Member verified.")
                temp=member
                break
        if temp is None:
            print(f"{borrower.name} is not a member.")
            return

        for book in temp.borrowed:
            if book.name.lower()==bookName.lower():
                book.borrowed=False
                book.borrower=None
                temp.borrowed.remove(book)
                print(f"{bookName} has been returned by {temp.name}")
                return

        print("Book does not exist in our database.")

    def searchBook(self,bookName):
        found=False
        for book in self.books:
            if book.name.lower()==bookName.lower():
                found=True
                print(f"Title : {book.name}\nAuthor : {book.author}\nBook ID : {book.book_id}\nCopy ID : {book.copy_id}\nBorrower : {'None' if book.borrower is None else book.borrower.name}")

        if not found:
            print("Book not found.")

    def searchMember(self,user):
        for member in self.members:
            if user.name.lower()==member.name.lower() and user.mem_id==member.mem_id:
                print(f"Name : {member.name}\nMember ID : {member.mem_id}\n Borrowed books :")
                for book in member.borrowed:
                    print(f"Title : {book.name}\tAuthor : {book.author}\tBook ID : {book.book_id}\tCopy ID : {book.copy_id}")
            return

        print("Member not found.")

    def view_books(self):
        if not self.books:
            print("No books in the library.")
            return
        for book in self.books:
            borrower_name = "None" if book.borrower is None else book.borrower.name
            print(f"{book.name} by {book.author} | Book ID: {book.book_id} | Copy ID: {book.copy_id} | Borrower: {borrower_name}")

    def view_members(self):
        if not self.members:
            print("No members registered.")
            return
        for member in self.members:
            print(f"{member.name} | Member ID: {member.mem_id} | Borrowed books: {len(member.borrowed)}")
    


library = Library()

while True:
    userInput = input("\nEnter a command (add_book/del_book/add_member/del_member/borrow/return/search_book/search_member/view_books/view_members/quit): ").strip().lower()

    match userInput:
        case "add_book":
            name = input("Book title: ")
            author = input("Author: ")
            book_id = input("Book ID: ")
            copy_id = input("Copy ID: ")
            library.add_book(Book(name, author, book_id, copy_id))

        case "del_book":
            book_id = input("Book ID: ")
            copy_id = input("Copy ID: ")
            temp = Book("", "", book_id, copy_id)
            library.del_book(temp)

        case "add_member":
            name = input("Member name: ")
            mem_id = input("Member ID: ")
            library.add_member(Member(name, mem_id))

        case "del_member":
            name = input("Member name: ")
            mem_id = input("Member ID: ")
            temp = Member(name, mem_id)
            library.del_member(temp)

        case "borrow":
            bookName = input("Book title to borrow: ")
            name = input("Your name: ")
            mem_id = input("Your member ID: ")
            borrower = Member(name, mem_id)
            library.borrow(bookName, borrower)

        case "return":
            bookName = input("Book title to return: ")
            name = input("Your name: ")
            mem_id = input("Your member ID: ")
            borrower = Member(name, mem_id)
            library.returnBook(bookName, borrower)

        case "search_book":
            bookName = input("Book title to search: ")
            library.searchBook(bookName)

        case "search_member":
            name = input("Member name: ")
            mem_id = input("Member ID: ")
            user = Member(name, mem_id)
            library.searchMember(user)

        case "view_books":
             library.view_books()

        case "view_members":
            library.view_members()

        case "quit":
            print("Exiting library system.")
            break

        case _:
            print("Unknown command.")
```
