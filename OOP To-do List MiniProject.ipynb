```python
class Task:
    def __init__(self, name, description):
        self.name = name
        self.desc = description
        self.completed = False

    def complete(self):
        self.completed = True

    def __repr__(self):
        status = "✓" if self.completed else "✗"
        return f"[{status}] {self.name}: {self.desc}"


class TaskList:
    def __init__(self):
        self.list = []

    def add(self, task):
        self.list.append(task)

    def remove(self, name):
        for task in self.list:
            if task.name == name:
                self.list.remove(task)
                return
        print("Task not found in list")

    def complete(self, name):
        for task in self.list:
            if task.name == name:
                task.complete()
                return
        print("Task not found in list")

    def display(self):
        for task in self.list:
            print(task)

l1=TaskList()
while True:
    userInput=input("Enter a command (add/remove/complete/display/quit): ")
    match userInput :
        case "add":
            name=input("Enter name of task.")
            desc=input("Enter a description for the task")
            l1.add(Task(name,desc))

        case "remove":
            name=input("Which Task do you want to remove ? ")
            l1.remove(name)

        case "complete":
            name=input("Which task do you want to mark complete ? ")
            l1.complete(name)

        case "display":
            l1.display()
            print("----------")
            print()

        case "quit":
            break

        case _:
            print("Wrong command!!!!!!!!")
        

    
```
