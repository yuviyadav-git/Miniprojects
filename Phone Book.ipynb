A simple contact book with JSON save/load

```python
import json

class Contact:
    def __init__(self,name,number):
        self.name=name
        self.number=number

class Phonebook:
    def __init__(self):
        self.book=[]
        self.load()

    def addContact(self,newContact):
        for cont in self.book:
            if len(cont.number) != 10:
                print("a number must be exactly of 10 digits.")
            else :
                if cont.name.lower() ==newContact.name.lower() and cont.number==newContact.number:
                    print("Contact already exists")
                    return
                self.book.append(newContact)
        

    def delContact(self,name,num):
        for cont in self.book:
            if cont.name.lower() ==name.lower() and cont.number==num:
                print("Contact deleted")
                self.book.remove(cont)
                return
        print("Contact not found in the Phonebook.")

    def search(self,oldContact):
        for cont in self.book:
            if cont.name.lower() ==oldContact.lower():
                print(f"Contact found!!\n{cont.name} : {cont.number}")
                return
        print("Contact not found.")

    def view(self):
        if not self.book:
            print("Phonebook is empty.")
            return
        for cont in self.book:
            print(f"{cont.name} : {cont.number}")

    def save(self):
        data=[{"name":c.name, "number":c.number} for c in self.book]
        with open("contacts.json", "w") as f:
            json.dump(data,f,indent=2)

    def load(self):
        try:
            with open("contacts.json","r") as f:
                    data=json.load(f)
            self.book=[Contact(d["name"],d["number"]) for d in data]
        except FileNotFoundError:
            self.book=[]


p1=Phonebook()

while True:
    command=input("Enter a command (add/remove/search/view/quit)")
    match command:
        case "add":
            name=input("Enter a name : ")
            num=input("Enter a number : ")
            p1.addContact(Contact(name,num))
            p1.save()

        case 'remove':
            name=input("Enter a name : ")
            num=input("Enter a number : ")
            p1.delContact(name,num)

        case 'search':
            name=input("Enter a name : ")
            p1.search(name)

        case 'view':
            p1.view()

        case 'quit':
            break

        case _:
            print("Invalid Input")
```
