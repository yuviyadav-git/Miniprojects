```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
max=5

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
sentences=[]
embeddings=[]
for i in range(0,max):
  sentence=input("Enter a sentence : ")
  sentences.append(sentence)
  embeddings.append(model.encode(sentence))
```

```python
embedding_mat=[]

def similarity(v1,v2):
  return round(float(np.dot(v1,v2)/(np.linalg.norm(v1)*np.linalg.norm(v2))),2)

for i in range(0,max):
  mat=[]
  for j in range(0,max):
    mat.append(similarity(embeddings[i],embeddings[j]))

  embedding_mat.append(mat)

for mat in embedding_mat:
  print(mat)
```

```python
E = embeddings
E_norm = E / np.linalg.norm(E, axis=1, keepdims=True)
sim_matrix = E_norm @ E_norm.T

print(np.round(sim_matrix, 2))
```
